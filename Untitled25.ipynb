```python
from pathlib import Path
import shutil

BASE = Path("/content/final_dataset")

# Clean old incomplete dataset
if BASE.exists():
    shutil.rmtree(BASE)

# Create correct folders
for folder in [
    BASE / "images" / "train",
    BASE / "images" / "val",
    BASE / "labels" / "train",
    BASE / "labels" / "val"
]:
    folder.mkdir(parents=True, exist_ok=True)

pascal = Path("/content/Some_Pascal-1")
mask = Path("/content/Face-Mask-18")

# Pascal: old class ID -> final class ID
pascal_map = {
    1: 2,    # bicycle -> Bike
    5: 3,    # bus -> Bus
    6: 1,    # car -> Car
    7: 5,    # cat -> Cat
    11: 4,   # dog -> Dog
    14: 0    # person -> Person
}

# Mask: old class ID -> final class ID
mask_map = {
    0: 6,    # Masque -> Mask
    1: 7,    # Notcorrect -> No_Mask
    2: 7     # PasMasque -> No_Mask
}

def merge_split(source, source_split, final_split, class_map, prefix):
    image_dir = source / source_split / "images"
    label_dir = source / source_split / "labels"

    count = 0

    for img in image_dir.iterdir():
        if img.suffix.lower() not in [".jpg", ".jpeg", ".png", ".bmp", ".webp"]:
            continue

        label = label_dir / f"{img.stem}.txt"

        if not label.exists():
            continue

        new_lines = []

        with open(label, "r") as f:
            for line in f:
                parts = line.strip().split()

                if len(parts) != 5:
                    continue

                old_id = int(parts[0])

                if old_id in class_map:
                    new_id = class_map[old_id]
                    new_lines.append(
                        f"{new_id} {' '.join(parts[1:])}\n"
                    )

        if not new_lines:
            continue

        new_img_name = f"{prefix}_{img.name}"
        new_label_name = f"{prefix}_{img.stem}.txt"

        shutil.copy2(
            img,
            BASE / "images" / final_split / new_img_name
        )

        with open(
            BASE / "labels" / final_split / new_label_name,
            "w"
        ) as f:
            f.writelines(new_lines)

        count += 1

    return count


# Pascal
pascal_train = merge_split(
    pascal, "train", "train", pascal_map, "pascal_train"
)

pascal_val = merge_split(
    pascal, "valid", "val", pascal_map, "pascal_val"
)

# Face Mask
mask_train = merge_split(
    mask, "train", "train", mask_map, "mask_train"
)

mask_val = merge_split(
    mask, "valid", "val", mask_map, "mask_val"
)

print("✅ MERGE COMPLETE")
print("Pascal train:", pascal_train)
print("Pascal val:", pascal_val)
print("Mask train:", mask_train)
print("Mask val:", mask_val)
```

```python
from pathlib import Path

yaml_content = """path: /content/final_dataset
train: images/train
val: images/val

names:
  0: Person
  1: Car
  2: Bike
  3: Bus
  4: Dog
  5: Cat
  6: Mask
  7: No_Mask
"""

Path("/content/final_dataset/data.yaml").write_text(yaml_content)

print("✅ data.yaml created successfully!")
print(yaml_content)
```
